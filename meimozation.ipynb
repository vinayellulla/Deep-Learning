```python
def fib(n):
  if n==1 or n==0:
    return 1
  else:
    return fib(n-1) + fib(n-2)
```

```python
fib(12)
```

```python
import time
start = time.time()
fib(36)
print(time.time()-start)
```

```python
start = time.time()
fib(40)                 # Exponential Complexcity
total_time = print(time.time()-start)
```

```python
import time
import matplotlib.pyplot as plt

n_values = []
times = []

for n in range(5, 41, 5):
    start = time.time()

    fib(n)

    total_time = time.time() - start

    n_values.append(n)
    times.append(total_time)

plt.plot(n_values, times, marker='o')

plt.xlabel("n")
plt.ylabel("Execution Time (seconds)")
plt.title("Fibonacci Execution Time")
plt.show()
```

```python
def fibb(n,d):
  if n in d:
    return d[n]
  else:
    d[n]= fibb(n-1,d) + fibb(n-2,d)
  return d[n]
```

```python
d={0:1,1:1}
fibb(400,d)
```

```python
import time
import matplotlib.pyplot as plt

n_values = []
times = []

for n in range(5, 101,5):
    d = {0: 0, 1: 1}

    start = time.time()

    fibb(n, d)

    total_time = time.time() - start

    n_values.append(n)
    times.append(total_time)

plt.plot(n_values, times, marker='o')

plt.xlabel("n")
plt.ylabel("Execution Time (seconds)")
plt.title("Fibonacci with Memoization")
plt.show()
```
